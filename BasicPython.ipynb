## Control Flow In Python

### Conditional Statements

```python
age = int(input("Enter your age : "))

if age >= 18:
   print("Your are eligible to vote")
   print("Thank you")

print("Succesfully logged out")   
```

### Nested if Statement

```python
personage = int(input("Enter your age :"))
city = input("Enter your city :")

if personage >= 18:
    if city == "Nagpur":
        print("Your are eligible to vote")
        print("Thank you")

print("Succesfully logged out")
```

### If Else Statement

```python
marks = int(input("Enter your marks :"))

if marks >= 40:
    print("You have passed your exam")
else:
    print("You have failed your exam")
```

### Nested If Else Statement

```python
n1 = int(input("Enter first number :"))
n2 = int(input("Enter second number :"))

if n1 >= n2:
    if n1 == n2:
        print (f'{n1} is equal to {n2}')
    else:
        print (f'{n1} is greater than {n2}')

else:
    print(f'{n1} is less than {2}')
```

# NESTED LOOP

```python
for i in range (3):
    print("Outer loop executed")
    for j in range (3):
        print("Inner loop executed")
        print(f'i = {i},j = {j}')

print("Out of loop")
```
